# Importing NYU Arrest Data and converting it to readable csv file

# Looking at the above simple codes, I can see my dataset has 19 columns with ~69K rows data. For this assignment I will be looking at the number of roberries reported in Brooklyn vs Manhattan commited by females to see which borough females commit more robberies in. Will also breakout by race 

```python
df2=df[df['OFNS_DESC'].str.contains('ROBBERY') & (df['ARREST_BORO'].isin(['M', 'B'])) & (df['PERP_SEX'] =='F')]
```

```python
df2
```

```python
df2.shape
```

# will be removing some columns that I dont belive bring much volume with what I'm looking into 

```python
drop_cols = [ 'LAW_CODE', 'LAW_CAT_CD','ARREST_PRECINCT', 'JURISDICTION_CODE', 'X_COORD_CD', 'Y_COORD_CD', 'Latitude', 'Longitude', 'Location']
```

```python
df2.drop(drop_cols, inplace = True, axis=1)
```

```python
df2.head(15)
```

```python
df2.shape
```

# Comparing Brooklyn vs Manhattan Female Robberies 

```python
robbery_boro = df2['ARREST_BORO'].value_counts()
```

```python
print(robbery_boro)
```

```python
robbery_boro.plot(kind='bar')
plt.title("Female Robbery Arrests: BK vs. Manhattan")
plt.xlabel("Borough")
plt.ylabel("Number of Arrests")
plt.show()
```

# Breakdown by Race 

```python
race_breakdown = (df2['PERP_RACE'].dropna().str.strip().value_counts().sort_values(ascending=False))
```

```python
print(race_breakdown)
```

```python
race_breakdown.plot(kind='bar')
plt.title("Female Robbery Arrests by Race")
plt.xlabel("Race")
plt.ylabel("Number of Arrests")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Breakdown of Race + Borough 

```python
race_boro_table = pd.crosstab(df2['ARREST_BORO'],df2['PERP_RACE'])
```

```python
print(race_boro_table)
```

```python
race_boro_table.plot(kind='bar', stacked=True)
plt.title("Female Robberies by Race & Borough")
plt.xlabel("Arrest Borough")
plt.ylabel("Number of Arrests")
plt.legend(title="Race")
plt.show()
```

# The above shows that female robberies are more prevalent in Brooklyn over Manhattan and are more often commited by Black & White Hispanics. In additon I would like to see how this breaks down by age  

```python
age_race_boro = df2.groupby(['ARREST_BORO', 'PERP_RACE', 'AGE_GROUP']).size().unstack(fill_value=0)
```

```python
print(age_race_boro)
```

```python
age_race_boro.groupby('ARREST_BORO').sum().plot(kind='bar', stacked=True)
plt.title("Female Robberies by Borough (All Ages & Races Combined)")
plt.xlabel("Borough")
plt.ylabel("Number of Arrests")
plt.xticks(rotation=0)
plt.tight_layout
plt.show()
```

# The above is a bit difficult to read/understand so breaking it out by age & borough instead below. But by just taking a look at the table that generate this bar graph, we can see that majority of these robberies are committed by females between the ages of 25-44 
 

```python
age_boro = pd.crosstab (df2['ARREST_BORO'], df2['AGE_GROUP'])
```

```python
print(age_boro)
```

```python
age_boro.plot(kind='bar')
plt.title("Female Robbery Arrests by Age Group & Borough")
plt.xlabel("Borough")
plt.ylabel("Number of Arrests")
plt.xticks(rotation=0)
plt.legend(title="Age Group")
plt.tight_layout()
plt.show()
```

# In conclusion, after reviewing and playing around with the data, one can come to the conclusion that Brooklyn has more female robberies over the borough of Manhattan. These robberies are often times commited more by Black and White Hispanice females within the age group of 25-44, closely followed by the age group of 18-24
